## 1. Import libraries and load the dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files

sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['axes.titlesize'] = 16
plt.rcParams['axes.labelsize'] = 12

# Upload the CSV file from your computer
uploaded = files.upload()
file_name = next(name for name in uploaded if name.lower().endswith('.csv'))
    
df = pd.read_csv(file_name)

print('Dataset shape:', df.shape)
display(df.head())
```

## 2. Basic data checking and cleaning

```python
# Check data types, missing values and duplicate rows
print('Missing values before cleaning:', df.isna().sum().sum())
print('Duplicate rows before cleaning:', df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates().copy()

# Convert numeric columns to numeric where applicable
numeric_cols = df.select_dtypes(include=np.number).columns
df[numeric_cols] = df[numeric_cols].apply(pd.to_numeric, errors='coerce')

# Fill missing numeric values with median
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())

    # Fill missing categorical values with mode
categorical_cols = df.select_dtypes(include='object').columns
for col in categorical_cols:
    if df[col].isna().any():
        df[col] = df[col].fillna(df[col].mode()[0])

print('Dataset shape after cleaning:', df.shape)
print('Missing values after cleaning:', df.isna().sum().sum())
```

# TOP 5 PROJECT VISUALIZATIONS

These five charts directly support the project's focus on seasonal patterns, agricultural performance, resource usage, environmental relationships and economic outcomes.

## Visualization 1 — Average Yield by Season
**Question:** How does agricultural yield vary across Kharif, Rabi and Zaid seasons?

**Why it is important:** Yield is a central measure of agricultural performance. Comparing average yield by season immediately shows which season performs better.

```python
season_order = ['Kharif', 'Rabi', 'Zaid']
season_yield = df.groupby('Season', as_index=False)['Yield_Tonnes_Ha'].mean()
season_yield['Season'] = pd.Categorical(season_yield['Season'], categories=season_order, ordered=True)
season_yield = season_yield.sort_values('Season')

plt.figure(figsize=(9, 6))
ax = sns.barplot(data=season_yield, x='Season', y='Yield_Tonnes_Ha', errorbar=None)
plt.title('Average Agricultural Yield by Season', fontweight='bold')
plt.xlabel('Season')
plt.ylabel('Average Yield (Tonnes/Ha)')
for container in ax.containers:
    ax.bar_label(container, fmt='%.2f', padding=3)
plt.tight_layout()
plt.show()
```

**Short result explanation:** In this dataset, Kharif has the highest average yield (about 5.64 tonnes/ha), followed by Rabi (about 5.08) and Zaid (about 4.67). This indicates a clear seasonal difference in agricultural productivity.

## Visualization 2 — Average Profit by Season
**Question:** Which season gives the strongest average economic performance?

**Why it is important:** Profit connects production performance with costs and revenue, making it useful for evidence-based agricultural planning.

```python
season_profit = df.groupby('Season', as_index=False)['Profit_INR'].mean()
season_profit['Season'] = pd.Categorical(season_profit['Season'], categories=season_order, ordered=True)
season_profit = season_profit.sort_values('Season')

plt.figure(figsize=(9, 6))
ax = sns.barplot(data=season_profit, x='Season', y='Profit_INR', errorbar=None)
plt.title('Average Profit by Agricultural Season', fontweight='bold')
plt.xlabel('Season')
plt.ylabel('Average Profit (INR)')
for container in ax.containers:
    ax.bar_label(container, fmt='%.0f', padding=3)
plt.ticklabel_format(style='plain', axis='y')
plt.tight_layout()
plt.show()
```

**Short result explanation:** This chart compares the economic outcome of the three seasons using average profit. It helps identify whether the season with better production performance also provides better financial performance.

## Visualization 3 — Average Yield by Crop
**Question:** Which crops contribute the highest average yield?

**Why it is important:** Crop-level comparison identifies high-performing and lower-performing crops and adds useful detail beyond the overall seasonal comparison.

```python
crop_yield = (df.groupby('Crop', as_index=False)['Yield_Tonnes_Ha']
             .mean()
             .sort_values('Yield_Tonnes_Ha', ascending=True))

plt.figure(figsize=(10, 7))
ax = sns.barplot(data=crop_yield, x='Yield_Tonnes_Ha', y='Crop', errorbar=None)
plt.title('Average Yield by Crop', fontweight='bold')
plt.xlabel('Average Yield (Tonnes/Ha)')
plt.ylabel('Crop')
for container in ax.containers:
    ax.bar_label(container, fmt='%.2f', padding=3)
plt.tight_layout()
plt.show()
```

**Short result explanation:** Sugarcane has a much higher average yield than the other crops in this dataset, while Pulses has the lowest average yield. This shows that crop type is an important dimension for understanding agricultural performance.

## Visualization 4 — Water Efficiency by Irrigation Method
**Question:** How does water-use efficiency differ across irrigation methods?

**Why it is important:** The project specifically asks for examination of resource usage. Water efficiency shows how much production is obtained per 1,000 m³ of water used.

```python
water_eff = (df.groupby('Irrigation_Method', as_index=False)
            ['Water_Efficiency_t_per_1000m3'].mean()
            .sort_values('Water_Efficiency_t_per_1000m3', ascending=False))

plt.figure(figsize=(10, 6))
ax = sns.barplot(data=water_eff, x='Irrigation_Method',
                y='Water_Efficiency_t_per_1000m3', errorbar=None)
plt.title('Average Water-Use Efficiency by Irrigation Method', fontweight='bold')
plt.xlabel('Irrigation Method')
plt.ylabel('Tonnes per 1,000 m³ of Water')
for container in ax.containers:
    ax.bar_label(container, fmt='%.2f', padding=3)
plt.tight_layout()
plt.show()
```

**Short result explanation:** Rainfed and Drip irrigation show higher average water-use efficiency than Flood irrigation in this dataset. This visualization supports discussion of resource-efficient agricultural practices.

## Visualization 5 — Rainfall vs Yield
**Question:** Is rainfall strongly associated with agricultural yield?

**Why it is important:** The project asks whether environmental conditions are related to agricultural outcomes. A scatter plot is appropriate for examining the relationship between two numeric variables.

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Rainfall_mm', y='Yield_Tonnes_Ha',
                hue='Season', alpha=0.65)
sns.regplot(data=df, x='Rainfall_mm', y='Yield_Tonnes_Ha',
            scatter=False, color='black', line_kws={'linewidth': 2})
plt.title('Rainfall vs Agricultural Yield', fontweight='bold')
plt.xlabel('Rainfall (mm)')
plt.ylabel('Yield (Tonnes/Ha)')
plt.legend(title='Season')
plt.tight_layout()
plt.show()

corr = df['Rainfall_mm'].corr(df['Yield_Tonnes_Ha'])
print(f'Pearson correlation between rainfall and yield: {corr:.3f}')
```

**Short result explanation:** The dataset shows a very weak linear relationship between rainfall and yield (correlation ≈ 0.031). Therefore, rainfall alone does not explain much of the variation in yield in this dataset; other agricultural, environmental and crop-related factors should also be considered.
